```python
import matplotlib.pyplot as plt
import numpy as np
from datetime import datetime

# ---------------------- 中文兼容配置 消除字体警告 ----------------------
plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False

# ---------------------- 1. 准备项目进度数据 ----------------------
tasks = ["制定计划", "方案设计", "资源调配", "第一阶段", "第二阶段", "第三阶段", "项目总结"]
start_dates = [
    datetime(2022,3,1), datetime(2022,3,13), datetime(2022,3,22),
    datetime(2022,4,2), datetime(2022,4,16), datetime(2022,5,11), datetime(2022,5,26)
]
progress = [0.51, 0.32, 0.21, 0.85, 0.36, 0.68, 0.68]
total_days = [11, 8, 10, 13, 24, 14, 7]
y = np.arange(len(tasks))
bar_height = 0.55

# ---------------------- 2. 画布与背景风格设置 ----------------------
plt.figure(figsize=(15, 7), dpi=120)
plt.gcf().set_facecolor('#161a45')
ax = plt.gca()
ax.set_facecolor('#161a45')

# ---------------------- 3. 绘制蓝青渐变甘特进度条 ----------------------
for idx, (s_date, prog, days) in enumerate(zip(start_dates, progress, total_days)):
    # 生成从亮青到深蓝的水平渐变
    grad = np.linspace(0, 1, 256).reshape(1, 256)
    bar_length = days * prog
    ax.imshow(grad, extent=[s_date.toordinal(), s_date.toordinal()+bar_length, 
                            y[idx]-bar_height/2, y[idx]+bar_height/2],
              aspect='auto', cmap='Blues_r', vmin=0.1, vmax=0.8)
    # 标注进度百分比
    ax.text(s_date.toordinal() + bar_length/2, y[idx],
            f'{int(prog*100)}%', ha='center', va='center', color='white', fontsize=11)

# ---------------------- 4. 坐标轴与时间刻度设置 ----------------------
# 自定义x轴时间标签
xtick_dates = [
    datetime(2022,3,1), datetime(2022,3,16), datetime(2022,3,31),
    datetime(2022,4,15), datetime(2022,4,30), datetime(2022,5,15),
    datetime(2022,5,30), datetime(2022,6,14)
]
ax.set_xticks([d.toordinal() for d in xtick_dates])
ax.set_xticklabels([d.strftime('%Y/%m/%d') for d in xtick_dates], color='white', fontsize=11)
# 绘制垂直参考虚线
for d in xtick_dates:
    ax.axvline(x=d.toordinal(), color='#666b8f', linestyle='--', linewidth=1.2, alpha=0.7)

ax.set_yticks(y)
ax.set_yticklabels(tasks, color='white', fontsize=12)
# 反转y轴，让任务顺序和原图完全匹配
ax.invert_yaxis()
# 隐藏所有四周边框
for spine in ax.spines.values():
    spine.set_visible(False)

# ---------------------- 5. 标题设置 ----------------------
plt.title('2022年化妆品类目采购项目进度', color='white', fontsize=20, fontweight='bold', pad=25)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.interpolate import make_interp_spline

# ---------------------- 中文兼容配置 消除字体警告 ----------------------
plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False

# ---------------------- 1. 准备月度销量数据 ----------------------
months = ["5月", "6月", "7月", "8月", "9月", "10月", "11月", "12月", "1月", "2月", "3月"]
sales = [146, 198, 296, 412, 506, 615, 789, 1021, 3782, 3215, 2936]
x = np.arange(len(months))

# ---------------------- 2. 生成平滑曲线 ----------------------
x_smooth = np.linspace(x.min(), x.max(), 300)
spl = make_interp_spline(x, sales, k=3)
y_smooth = spl(x_smooth)

# ---------------------- 3. 画布与背景风格设置 ----------------------
plt.figure(figsize=(14, 7.5), dpi=120)
plt.gcf().set_facecolor('#161a45')
ax = plt.gca()
ax.set_facecolor('#161a45')

# ---------------------- 4. 绘制橙色平滑趋势线 ----------------------
ax.plot(x_smooth, y_smooth, color='#d9664f', linewidth=2.2)
# 标注峰值数值 + 峰值虚线
peak_idx = sales.index(max(sales))
ax.axvline(x=peak_idx, color='#d9664f', linestyle='--', linewidth=1.5, alpha=0.7)
ax.text(peak_idx, max(sales)+220, f'{max(sales)}', ha='center', va='bottom', color='white', fontsize=13)

# ---------------------- 5. 坐标轴与样式设置 ----------------------
ax.set_ylim(0, 4200)
ax.set_yticks([0, 1000, 2000, 3000, 4000])
ax.tick_params(axis='y', colors='#d9664f', labelsize=11)
ax.set_xticks(x)
ax.set_xticklabels(months, color='white', fontsize=12)
# 隐藏四周边框，仅保留底部浅色基线
for spine in ax.spines.values():
    spine.set_visible(False)
ax.axhline(y=0, color='white', linewidth=1.2)

# ---------------------- 6. 底部年份标签条 ----------------------
# 2021年青色标签
ax.add_patch(plt.Rectangle((0, -750), 7.8, 450, color='#66ccdd', ec='none', zorder=0))
ax.text(3.8, -520, '2021', ha='center', va='center', color='white', fontsize=14, fontweight='medium')
# 2022年黄色标签
ax.add_patch(plt.Rectangle((7.9, -750), 3.1, 450, color='#f7c04a', ec='none', zorder=0))
ax.text(9.5, -520, '2022', ha='center', va='center', color='white', fontsize=14, fontweight='medium')

# ---------------------- 7. 标题与底部注释 ----------------------
plt.text(-0.3, 3900, '化妆品品类月度销量走势', color='white', fontsize=28, fontweight='bold')
plt.text(-0.3, 3400, '2022年销量迅速增加，1月最高，销量达到3782', color='white', fontsize=19)
plt.text(-0.3, -1300, '*注：数据来源于公司销售系统，统计日期截至2022.03.31', color='white', fontsize=11)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# ---------------------- 中文兼容配置 消除字体警告 ----------------------
plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False

# ---------------------- 1. 准备完成率数据 ----------------------
complete_rate = 0.65

# ---------------------- 2. 画布与背景风格设置 ----------------------
plt.figure(figsize=(10, 7), dpi=120)
plt.gcf().set_facecolor('#161a45')
ax = plt.gca()
ax.set_facecolor('#161a45')

# ---------------------- 3. 绘制半填充圆形进度 ----------------------
# 生成下半部分填充的蓝色半圆
theta = np.linspace(np.pi, 2*np.pi, 200)
x_fill = np.cos(theta)
y_fill = np.sin(theta)
# 调整填充高度匹配65%完成率
y_fill = y_fill * complete_rate * 2 - (1 - complete_rate)
ax.fill(x_fill, y_fill, color='#0072ce', zorder=2)

# 绘制双层蓝色外圆环
circle_outer = plt.Circle((0, 0), 1.1, color='#0072ce', fill=False, linewidth=2.5, zorder=3)
circle_inner = plt.Circle((0, 0), 0.9, color='#0072ce', fill=False, linewidth=2.5, zorder=3)
ax.add_patch(circle_outer)
ax.add_patch(circle_inner)

# ---------------------- 4. 中心百分比文字标注 ----------------------
ax.text(0, 0, f'{int(complete_rate*100)}%', ha='center', va='center', 
        color='white', fontsize=42, fontweight='medium')

# ---------------------- 5. 坐标轴隐藏 ----------------------
ax.set_xlim(-1.6, 1.6)
ax.set_ylim(-1.6, 1.6)
ax.set_xticks([])
ax.set_yticks([])
for spine in ax.spines.values():
    spine.set_visible(False)

# ---------------------- 6. 标题与底部注释 ----------------------
plt.text(-1.5, 1.3, '2022年上半年目标完成率', color='white', fontsize=22, fontweight='bold')
plt.text(-1.5, 1.1, '截至6月30日销售目标总体完成率达到65%', color='white', fontsize=15)
plt.text(-1.5, -1.45, '*注：数据来源于公司销售系统', color='white', fontsize=10)

# 保证图形为正圆形
ax.axis('equal')
plt.tight_layout()
plt.show()
```
